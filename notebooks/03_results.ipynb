Out-of-sample (2025-26) comparison of the four strategies. Generated tables/figures come from `scripts/03_run_backtests.py`; this notebook reloads and inspects them.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))
import pandas as pd

RESULTS = pathlib.Path('..') / 'results' / 'nba_2025-26'
summary = pd.read_csv(RESULTS / 'summary.csv', index_col=0)
summary[['total_pnl_usd','mean_pnl_per_game_usd','game_sharpe','win_rate',
         'total_fees_usd','spread_captured_usd','inventory_pnl_usd',
         'mean_abs_inv','mean_terminal_abs_inv','fills_per_game']].round(2)
```

```python
from IPython.display import Image, display
for f in ['cumulative_pnl.png','decomposition.png','inventory.png','pnl_distribution.png']:
    display(Image(str(RESULTS / f)))
```

```python
# PnL vs hours-to-tip exposure: which strategies survive the final hour?
import matplotlib.pyplot as plt
per_game = {p.stem.replace('per_game_',''): pd.read_parquet(p) for p in RESULTS.glob('per_game_*.parquet')}
fig, ax = plt.subplots(figsize=(9,4))
for name, pg in per_game.items():
    ax.scatter(pg['tape_volume'], pg['pnl_c']/100, s=6, alpha=0.4, label=name)
ax.set_xscale('log'); ax.set_xlabel('game tape volume'); ax.set_ylabel('PnL (USD)')
ax.legend(); ax.set_title('Per-game PnL vs liquidity');
```
